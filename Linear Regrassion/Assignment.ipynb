## Q4. Data Preprocessing

    Prepare the dataset by:

    Separating features (X) and target (y)
    Splitting the data into 80% training and 20% testing sets

```python

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print(X_train.shape)

X_test.shape
```

## Q5. Model Training

## Build a Linear Regression model using the training dataset and display the model coefficients and intercept.

```python
model = LinearRegression()

model.fit(X_train, y_train)

print("\nModel Coefficients:")

for feature, coefficient in zip(X.columns, model.coef_):
    print(feature, ":", coefficient)

print("\nModel Intercept:")
print(model.intercept_)
```

## Q6. Prediction

## Predict the target values for the testing dataset and display a comparison of the Actual and Predicted values.

```python
y_pred = model.predict(X_test)

comparison = pd.DataFrame({
    "Actual": y_test.values,
    "Predicted": y_pred
})

print(comparison)
```

## Q7. Model Evaluation

        Evaluate the model using:

        Mean Absolute Error (MAE)
        Mean Squared Error (MSE)
        Root Mean Squared Error (RMSE)
        R² Score
        Interpret the results.

```python
import numpy as np
mae = mean_absolute_error(y_test, y_pred)

mse = mean_squared_error(y_test, y_pred)

rmse = np.sqrt(mse)

r2 = r2_score(y_test, y_pred)


print("MAE :", mae)
print("MSE :", mse)
print("RMSE:", rmse)
print("R2 Score:", r2)
```

## Interpretation

    MAE, MSE, and RMSE measure the prediction errors of the model. Lower values indicate smaller errors. R² indicates how much variation in disease progression is explained by the model. A higher R² indicates better explanatory performance.
